```python
import numpy as np
import matplotlib.pyplot as plt
import pybromo as pbm
import os, time, tables
# directory to save HDF5
folder_sims = './Pybromo_simulations'
folder_ts = './Pybromo_timestamps'
os.makedirs(folder_sims, exist_ok=True)
os.makedirs(folder_ts, exist_ok=True)
```

# I. **free EtBr** ts

```python
Sim_01 = pbm.ParticlesSimulation.from_datafile('dda833', path=folder_sims, ignore_timestamps=True)
N = Sim_01.num_particles  
rs_ts = np.random.RandomState()
rs_ts.set_state(Sim_01.traj_group._v_attrs['last_random_state'])
print(f'Sim 01 [EtBr]:\n- {Sim_01.box}\n- N particles: {Sim_01.num_particles}\n- Diffusion coeff. {Sim_01.diffusion_coeff[0]} m^2/s')
```

```python
# emission parameters: high limit max and background rates
max_rate = 600   # cps
bg_rate = 275     # cps

# generate timestamps of photons emitted from molecules diffusion
Sim_01.simulate_timestamps_mix(
    max_rates=(max_rate,) * N,
    populations=(slice(0, N),),
    bg_rate=bg_rate,
    rs=rs_ts,
    overwrite=True, 
    path = folder_ts
)
# get timestamps number information and print it
ph_times = Sim_01._timestamps.read() 
print(f'Total number of generated photons: {len(ph_times)}')
# verifica quali array timestamps esistono
Sim_01.timestamps_match_mix(
    max_rates=(max_rate, ) * N,
    populations=(slice(0, N),),
    bg_rate=bg_rate,
)
```

```python
Sim_01.ts_store.close()
```

# II. **EtBr-DNA** ts (0.12 blobs in confocal vol.)

```python
Sim_03 = pbm.ParticlesSimulation.from_datafile('8188ec', path=folder_sims, ignore_timestamps=True)
N_EtBr_free, N_EtBr_DNA = Sim_03.particles.particles_counts  
rs_ts = np.random.RandomState()
rs_ts.set_state(Sim_03.traj_group._v_attrs['last_random_state'])
D_coeffs = np.unique(Sim_03.diffusion_coeff)
print(f'Sim 03 [EtBr-DNA]:\n- {Sim_03.box}\n- N EtBr-free: {N_EtBr_free} | Diffusion coeff. {D_coeffs[1]} m^2/s')
print(f'- N EtBr-DNA: {N_EtBr_DNA}    | Diffusion coeff. {D_coeffs[0]} m^2/s')
```

```python
# timestamps: emission + background parameters
max_rate_free = 600     # cps, EtBr free
max_rate_dna  = 24000    # cps, EtBr-DNA
bg_rate       = 275 + 2400
# 2 populations, ordered like in simulation: free first, then DNA-bound
pop_free = slice(0, N_EtBr_free)
pop_dna  = slice(N_EtBr_free, N_EtBr_free + N_EtBr_DNA)
# recover diffusion random stream for reproducibility
rs_ts = np.random.RandomState()
rs_ts.set_state(Sim_03.traj_group._v_attrs['last_random_state'])
# UN max_rate per population (len(max_rates) == len(populations))
Sim_03.simulate_timestamps_mix(
    max_rates   = (max_rate_free, max_rate_dna), 
    populations = (pop_free, pop_dna),
    bg_rate     = bg_rate,
    rs          = rs_ts,
    overwrite   = True,
    path        = folder_ts,
)
# info generated timestamps
ph_times  = Sim_03._timestamps.read()
ph_par    = Sim_03._tparticles.read()
print(f'\nTotal number of generated photons: {len(ph_times)}')
print(f'Photons from population "free": {(ph_par < N_EtBr_free).sum()}')
print(f'Photons from population "DNA":  {(ph_par >= N_EtBr_free).sum()}')

# get timestamps metadata
Sim_03.timestamps_match_mix(
    max_rates=(max_rate_free, max_rate_dna),
    populations=(pop_free, pop_dna),
    bg_rate=bg_rate,
)
```

```python
Sim_03.ts_store.close()
```
