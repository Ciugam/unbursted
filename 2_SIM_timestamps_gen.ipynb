```python
import numpy as np
import matplotlib.pyplot as plt
import pybromo as pbm
import os, time, tables
# directory to save HDF5
folder_sims = './Pybromo_simulations'
folder_ts = './Pybromo_timestamps'
os.makedirs(folder_sims, exist_ok=True)
os.makedirs(folder_ts, exist_ok=True)
```

# I. **free EtBr** ts (100 molecules in the Box)

```python
Sim_free = pbm.ParticlesSimulation.from_datafile('ef0466', path=folder_sims, ignore_timestamps=True)
N = Sim_free.num_particles  
rs_ts = np.random.RandomState()
rs_ts.set_state(Sim_free.traj_group._v_attrs['last_random_state'])
print(f'Sim 01 [EtBr]:\n- {Sim_free.box}\n- N particles: {Sim_free.num_particles}\n- Diffusion coeff. {Sim_free.diffusion_coeff[0]} m^2/s')
```

```python
# emission parameters: high limit max and background rates
max_rate = 600   # cps
bg_rate = 275     # cps

# generate timestamps of photons emitted from molecules diffusion
Sim_free.simulate_timestamps_mix(
    max_rates=(max_rate,) * N,
    populations=(slice(0, N),),
    bg_rate=bg_rate,
    rs=rs_ts,
    overwrite=True, 
    path = folder_ts
)
# get timestamps number information and print it
ph_times = Sim_free._timestamps.read() 
print(f'Total number of generated photons: {len(ph_times)}')
# verifica quali array timestamps esistono
Sim_free.timestamps_match_mix(
    max_rates=(max_rate, ) * N,
    populations=(slice(0, N),),
    bg_rate=bg_rate,
)
```

```python
Sim_free.ts_store.close()
```

# II. **EtBr-DNA** ts (6 blobs in the Box)

```python
Sim_006 = pbm.ParticlesSimulation.from_datafile('767b38', path=folder_sims, ignore_timestamps=True)
N_EtBr_free, N_EtBr_DNA = Sim_006.particles.particles_counts  
rs_ts = np.random.RandomState()
rs_ts.set_state(Sim_006.traj_group._v_attrs['last_random_state'])
D_coeffs = np.unique(Sim_006.diffusion_coeff)
print(f'Sim 03 [EtBr-DNA]:\n- {Sim_006.box}\n- N EtBr-free: {N_EtBr_free} | Diffusion coeff. {D_coeffs[1]} m^2/s')
print(f'- N EtBr-DNA: {N_EtBr_DNA}    | Diffusion coeff. {D_coeffs[0]} m^2/s')
```

```python
# timestamps: emission + background parameters
max_rate_free = 600     # cps, EtBr free
max_rate_dna  = 10000    # cps, EtBr-DNA
bg_rate       = 275
# 2 populations, ordered like in simulation: free first, then DNA-bound
pop_free = slice(0, N_EtBr_free)
pop_dna  = slice(N_EtBr_free, N_EtBr_free + N_EtBr_DNA)
# recover diffusion random stream for reproducibility
rs_ts = np.random.RandomState()
rs_ts.set_state(Sim_006.traj_group._v_attrs['last_random_state'])
# UN max_rate per population (len(max_rates) == len(populations))
Sim_006.simulate_timestamps_mix(
    max_rates   = (max_rate_free, max_rate_dna), 
    populations = (pop_free, pop_dna),
    bg_rate     = bg_rate,
    rs          = rs_ts,
    overwrite   = True,
    path        = folder_ts,
)
# info generated timestamps
ph_times  = Sim_006._timestamps.read()
ph_par    = Sim_006._tparticles.read()
print(f'\nTotal number of generated photons: {len(ph_times)}')
print(f'Photons from population "free": {(ph_par < N_EtBr_free).sum()}')
print(f'Photons from population "DNA":  {(ph_par >= N_EtBr_free).sum()}')

# get timestamps metadata
Sim_006.timestamps_match_mix(
    max_rates=(max_rate_free, max_rate_dna),
    populations=(pop_free, pop_dna),
    bg_rate=bg_rate,
)
```

```python
Sim_006.ts_store.close()
```
